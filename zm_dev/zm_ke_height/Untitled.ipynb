```python
import seaborn as sns
import matplotlib.pyplot as plt

# Sample data
data = [1, 2, 2, 3, 3, 3, 4, 4, 5]

# Create the histogram
sns.histplot(data, kde=False, stat="density")

# Overlay the KDE plot
sns.kdeplot(data, color="red")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate sample data
np.random.seed(42)
data = np.random.normal(loc=0, scale=1, size=1000)

# Create a histogram with density on the y-axis
plt.figure(figsize=(8, 6))
plt.hist(data, bins=30, density=True, color='skyblue', edgecolor='black', alpha=0.7)

# Add titles and labels
plt.title("Histogram with Density on Y-axis", fontsize=16)
plt.xlabel("Data Values", fontsize=14)
plt.ylabel("Density", fontsize=14)

# Show the plot
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Generate sample data
np.random.seed(42)
data = np.random.normal(loc=0, scale=1, size=1000)

# Create a histogram with Seaborn
plt.figure(figsize=(8, 6))
sns.histplot(data, bins=30, kde=False, color='skyblue')

# Add titles and labels
plt.title("Histogram with Values on X-axis", fontsize=16)
plt.xlabel("Data Values", fontsize=14)
plt.ylabel("Frequency", fontsize=14)

# Show the plot
plt.tight_layout()
plt.show()
```
