```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Example altitudes in meters
altitude = np.linspace(0, 10000, 100)

# Two synthetic pressure profiles (e.g., different conditions or times)
pressure_profile1 = 1013.25 * np.exp(-altitude / 7000)  # exponential decrease
pressure_profile2 = 1013.25 * np.exp(-altitude / 8000)  # slower decrease
```

```python
plt.figure(figsize=(6, 8))

# Plot the pressure profiles
plt.plot(pressure_profile1, altitude, label='Profile 1')
plt.plot(pressure_profile2, altitude, label='Profile 2')

# Fill between the profiles
plt.fill_betweenx(altitude, pressure_profile1, pressure_profile2, color='gray', alpha=0.5)

# Labeling the axes
plt.xlabel('Pressure (hPa)')
plt.ylabel('Altitude (m)')
plt.title('Pressure Profiles and Range Between Them')
plt.legend()

# Inverting y-axis to have 0 at the bottom, higher altitude at the top
plt.gca().invert_yaxis()

plt.show()
```
