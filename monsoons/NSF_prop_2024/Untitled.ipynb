```python
import xarray as xr
import geopandas as gpd
import rioxarray
from shapely.geometry import mapping

# Load the NetCDF dataset
ds = xr.open_dataset('your_dataset.nc')

# Load the shapefile or geojson of the country boundary
country_shape = gpd.read_file('your_country_shapefile.shp')

# Reproject the shapefile to match the NetCDF dataset's CRS if needed
country_shape = country_shape.to_crs("EPSG:4326")

# Select the country boundary and create a mask
geometry = [mapping(geom) for geom in country_shape.geometry]
ds = ds.rio.write_crs("EPSG:4326")  # Assign CRS if not already set

# Apply the mask: set points outside the geometry to NaN
masked_ds = ds.rio.clip(geometry, ds.rio.crs, drop=True)

# Check results
print(masked_ds)
```
