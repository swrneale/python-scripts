```python

# These are here, but actually taken from ~/.cdsapirc
url = 'https://cds.climate.copernicus.eu/api/v2'
key: '98208:5398f2c2-bba0-472b-bc7e-8aa81b874a6d'

#dir_out  = '/glade/work/rneale/data/ERA5/from_cdsapi/'
dir_out = '/glade/scratch/rneale/ERA5/'

c = cdsapi.Client()

var_3d = 'dtdt_conv'  ; var_3d_name = 'tendency_of_air_temperature_due_to_convective_heating'


for iyr in range(1979,1980):

    prod_type = 'reanalysis-era5-model-levels'
    file_out = dir_out+var_3d+'_test_era5_hourly_'+str(iyr)+'.nc'
    
    ret_options = {
            'product_type': 'reanalysis',
            'variable': var_3d_name,
            'year': [str(iyr)],
            'model_level': ['60'],
            'month': ['01'],
            'day': ['01'],  # Add more days as needed
            'time': [f'{h:02d}:00' for h in range(24)],
            'area': [50, 0, 360, -50],  # North, West, South, East (e.g., US)
            'format': 'netcdf',
        }
       

    print(prod_type,ret_options,file_out)

    c.retrieve(prod_type,ret_options,file_out)



```
