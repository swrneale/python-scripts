# Code to grab Monthly Means ERA5 data using cdsapi

```python
import cdsapi
```

```python

# These are here, but actually taken from ~/.cdsapirc
url = 'https://cds.climate.copernicus.eu/api'
key: '30fcd54e-05d1-4d24-af2f-e27593f59ed8'

#dir_out  = '/glade/work/rneale/data/ERA5/from_cdsapi/'
dir_out = '/glade/derecho/scratch/rneale/ERA5/mmean/0.25deg/'

c = cdsapi.Client()

for iyr in range(1940,2024):

    c.retrieve(
        "reanalysis-era5-single-levels-monthly-means",
        {
            "product_type": ["monthly_averaged_reanalysis"],
            "variable": ["mean_eastward_turbulent_surface_stress"],
            "year": [
                str(iyr),
            ],
            "month": [
                "01", "02", "03",
                "04", "05", "06",
                "07", "08", "09",
                "10", "11", "12"],
            
            "time": ["00:00"],
            
            "data_format": "netcdf",
            "download_format" : "unarchived"
        },
        dir_out+'taux_era5_monthly_'+str(iyr)+'.nc')




```
