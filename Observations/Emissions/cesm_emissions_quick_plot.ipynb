```python
import xarray as xr
import matplotlib.pyplot as mp
```

```python
indir0 = '/glade/campaign/cesm/cesmdata/cseg/inputdata/atm/cam/ggas/'
infile = 'emissions-cmip6_CO2_anthro_surface_ScenarioMIP_IAMC-AIM-ssp370_175001-210112_fv_0.9x1.25_c20201210.nc'

print(indir0+infile)
ds_emis = xr.open_dataset(indir0+infile, engine="netcdf4")



dlat = 45.
dlon = 0.
#display(ds_emis)
da_emis = ds_emis['CO2_flux'].sel(lat=dlat,lon=dlon, method="nearest").squeeze()
da_ann = da_emis.groupby('time.year').mean('time')
display(da_ann)
da_ann.plot(figsize=(12, 5))
```
