```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

# Create dummy data
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Specify unevenly spaced contour intervals
contour_intervals = [-0.3, -0.25, -0.2, -0.1, 0, 0.2, 0.5, 0.7, 0.9, 1.0]

# Define custom colormap with 10 colors
colors = ['white', 'cyan', 'green', 'yellow', 'orange', 'red', 'pink', 'purple', 'brown', 'blue']
cmap = LinearSegmentedColormap.from_list('custom_colormap', colors, N=len(colors)-1)

# Create filled contour plot with 10 colors for the specified intervals
contour = plt.contourf(X, Y, Z, levels=contour_intervals, cmap=cmap, extend='both')

# Add colorbar
cbar = plt.colorbar(contour, ticks=contour_intervals)
cbar.set_label('Z values')

# Show the plot
plt.title('Filled Contour Plot with 10 Unevenly Spaced Contour Intervals and 10 Colors')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()


```
