```python
import xarray as xr
import numpy as np
import fsspec
import calendar
```

```python
s3_path = 's3://noaa-oar-sfsdev-pds/experiments/beta1/reforecast/03/atm_daily.zarr'
mapper = fsspec.get_mapper(s3_path, anon=True)
atm = xr.open_zarr(mapper, consolidated=True)
```

```python
print(atm.variables)
```

```python
prate = atm.PRATE_surface
print(prate)
print(prate[4,0,:,90,180].values*60*60*24)
```

```python
mon_mean_prate = prate[4,0,:,90,180]
print(mon_mean_prate.values)
days_in_month = []
for month in range(1,13):
    days_in_month.append(calendar.monthrange(1997, month)[1])
print(days_in_month)
annual_precip = np.sum(mon_mean_prate*60*60*24 * np.array(days_in_month))
print(annual_precip.values)
```
