```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
td0 = -10.0
alpha = 1.4
maxscl = 10.0

# Ranges
tdiff = np.linspace(-30, -5, 300)
vmag = np.linspace(0, 20, 300)

# 2-D grids
TD, VM = np.meshgrid(tdiff, vmag)

# Calculate vscl
vscl = np.minimum(
    1.0 + alpha * (np.abs(TD - td0)**0.5 / np.abs(VM)),
    maxscl
)

# Contour plot
fig, ax = plt.subplots(figsize=(9, 6))

levels = np.linspace(1.0, 10.0, 55)

cf = ax.contourf(TD, VM, vscl, levels=levels)
cs = ax.contour(TD, VM, vscl, levels=levels, colors='black', linewidths=0.5)

ax.clabel(cs, inline=True, fontsize=9, fmt='%.1f')

cbar = fig.colorbar(cf, ax=ax)
cbar.set_label('vscl')

ax.set_xlabel('tdiff (°C)')
ax.set_ylabel('vmag')
ax.set_title(
    f'vscl = min(1 + α√|tdiff − td0| / |vmag|, maxscl)\n'
    f'td0 = {td0}, α = {alpha}, maxscl = {maxscl}'
)

plt.tight_layout()
plt.show()
```
